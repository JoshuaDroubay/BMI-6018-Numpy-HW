```python
#3

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

column_names = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]

iris_df = pd.read_csv(url, header=None, names=column_names)

print(iris_df.head())
```

```python
#4

i = 0
while (iris_df.at[i, "petal_width"] <= 1) and i < iris_df.shape[0]:
    i = i + 1

problem4 = i
print(problem4)
```
